This notebook reproduces the schema, quality, leakage, crop, district, risk, and time-trend checks used by AgriShield AI. Run Jupyter from the project root.

```python
import pandas as pd
from src.config import RAW_DATA_PATH, TARGET
from src.data_processing import load_data
from src.eda import analyze_dataset
data = load_data(RAW_DATA_PATH)
summary = analyze_dataset(data)
summary
```

```python
data.describe(include='all').T
```

```python
data.groupby('crop')[TARGET].agg(['count', 'mean', 'median', 'std', 'min', 'max']).sort_values('mean', ascending=False)
```

```python
data.groupby('district').agg(mean_yield=(TARGET, 'mean'), mean_rainfall=('annual_rainfall_mm', 'mean'), mean_temperature=('avg_temperature_c', 'mean'), mean_risk=('overall_climate_risk_score', 'mean')).sort_values('mean_yield', ascending=False)
```
